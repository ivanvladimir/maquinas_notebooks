```python
import ipywidgets as widgets
from ipywidgets import interact, interact_manual
```

```python
from maquinas.io import load_jflap
from maquinas.collections import list_collections, get_collection
```

```python
dropdown = widgets.Dropdown(
    options=list_collections(),
    description="Colections:",
)
dropdown
```

```python
collection=get_collection(dropdown.value)
```

```python
@interact
def explore(name=collection.names()):
    m=collection.load(name)
    m.print_summary()
    return m.draw()
```
