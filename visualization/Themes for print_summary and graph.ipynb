```python
from maquinas.io import load_fa
from maquinas import themes
from maquinas import summary
from maquinas.summary import SummaryStyle
```

```python
m=load_fa("""
        | 1   | 0
-> q_1  | q_1 | q_2
   q_2] |     | q_2""")
```

```python
# Every theme, light and dark
for name, t in themes.THEMES.items():
    print(f"--- {name}")
    m.print_summary(style=t.style)

# Plus the built-in presets
for name, s in [("vibrant", SummaryStyle.vibrant()),
                ("white", SummaryStyle.white()),
                ("plain", SummaryStyle.plain())]:
    print(f"--- {name}")
    m.print_summary(style=s)
```

```python
from IPython.display import display
from maquinas.drawing import GraphStyle

for name, t in themes.THEMES.items():
    print(name)
    display(m.graph(style=t.style, graph_style=t.graph))

for name, gs in [("default", GraphStyle()), ("plain", GraphStyle.plain())]:
    print(name)
    display(m.graph(graph_style=gs))
```
