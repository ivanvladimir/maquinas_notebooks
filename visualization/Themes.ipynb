```python
from maquinas.io import load_fa, load_cfg
from maquinas import themes
from maquinas import summary
from maquinas.summary import SummaryStyle
```

```python
m=load_fa("""
        |  a       |  b  |  c  |
 ->q_0  | q_0, q_1 | q_0 | q_0 |
   q_1  | q_0      | q_2 | q_0 |
   q_2  | q_0      | q_0 | q_3 |
   q_3] | q_3      | q_3 | q_3 |
""")

g=load_cfg("""
S->aSb; S-> epsilon;
""")

# Analysis de cadena
roots,chart,forest=g.parse('aaaabbbb')
# Extrae árboles
trees=list(g.extract_trees(forest))
```

```python
# Every theme, light and dark
for name, t in themes.THEMES.items():
    print(f"--- {name}")
    m.print_summary(style=t.style)

# Plus the built-in presets
for name, s in [("vibrant", SummaryStyle.vibrant()),
                ("white", SummaryStyle.white()),
                ("plain", SummaryStyle.plain())]:
    print(f"--- {name}")
    m.print_summary(style=s)
```

```python
from IPython.display import display
from maquinas.drawing import GraphStyle

for name, t in themes.THEMES.items():
    print(name)
    display(m.draw(style=t.style, graph_style=t.graph))

for name, gs in [("default", GraphStyle()), ("plain", GraphStyle.plain())]:
    print(name)
    display(m.draw(graph_style=gs))
```

```python
for name, t in themes.THEMES.items():
    print(name)
    display(m.save_animation("abc",filename="maquina.svg",style=t.style, graph_style=t.graph,show=True))

for name, gs in [("default", GraphStyle()), ("plain", GraphStyle.plain())]:
    print(name)
    display(m.save_animation("abc",filename="maquina.svg",graph_style=gs,show=True))
```

```python
for name, t in themes.THEMES.items():
    print(name)
    display(g.draw_trees(trees,style=t.style, graph_style=t.graph))

for name, gs in [("default", GraphStyle()), ("plain", GraphStyle.plain())]:
    print(name)
    display(g.draw_trees(trees,graph_style=gs))
```
